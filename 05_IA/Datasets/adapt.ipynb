```python
import pandas as pd
```

```python
ibex_file_path = 'IBEX-1994-2020.csv'
file_path = 'DJ_sorted.csv'
ibex_df = pd.read_csv(ibex_file_path)
df = pd.read_csv(file_path)
```

```python
df['Date'] = pd.to_datetime(df['Date'], infer_datetime_format=True).dt.strftime('%Y-%m-%d')
```

```python
ibex_dates = set(ibex_df['Date'])
```

```python
rows_to_drop = []
for index, row in df.iterrows():
    if row['Date'] not in ibex_dates:
        rows_to_drop.append(index)
```

```python
df = df.drop(rows_to_drop)
```

```python
df = df[::-1].reset_index(drop=True)
```

```python
adapted_dow_jones_file_path = 'DJ_adapted.csv'
df.to_csv(adapted_dow_jones_file_path, index=False)
```
