```python
import pandas as pd
```

```python
file_path = 'DOW_JONES.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
df = df.drop(['Vol.', 'Change %'], axis=1)
df.head()
```

```python
df.columns = ['Date', 'DJ_close', 'DJ_open', 'DJ_high', 'DJ_low']
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

```python
numeric_cols = ['DJ_close', 'DJ_open', 'DJ_high', 'DJ_low']
for col in numeric_cols:
    # Remove periods (thousand separators)
    df[col] = df[col].str.replace(',', '')
    # Replace commas (decimal separators) with periods
    #df[col] = df[col].str.replace('.', '.')
    # Convert to float
    df[col] = df[col].astype(float)
df.head()
```

```python
df = df[['Date', 'DJ_open', 'DJ_high', 'DJ_low', 'DJ_close']]
df.head()
```

```python
df = df.sort_values(by='Date')
df.head()
```

```python
df.to_csv('DOW_JONES_2021.csv', index=False)
```
