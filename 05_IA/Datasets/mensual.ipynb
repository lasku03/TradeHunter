```python
import pandas as pd
import calendar
```

```python
xls_file_path = "datos.xlsx"
df = pd.read_excel(xls_file_path, header=None)
df.head(347)
```

```python
df = df.drop(0, axis=0)
df = df.drop(1, axis=0)
```

```python
dfs = []
df = pd.DataFrame(df)

# Elimina el % y el € de las columnas que contienen estos caracteres
df[2] = df[2].str.rstrip('%')
df[3] = df[3].str.rstrip('€')

df.head(347)
```

```python
for index, row in df.iterrows():
    monthYear = row[0]
    debt = row[1]
    percentage = row[2]
    debt_per_capita = row[3]
    year = monthYear.year
    month = monthYear.month

    is_leap_year = calendar.isleap(year)
    
    start_date = pd.to_datetime(f"{year}-{month}-01")

    if month == 4 or month == 6 or month == 9 or month == 11:
        end_date = pd.to_datetime(f'{year}-{month}-30')
    elif month == 2:
        if is_leap_year:
            end_date = pd.to_datetime(f'{year}-{month}-29')
        else:
            end_date = pd.to_datetime(f'{year}-{month}-28')
    else:
        end_date = pd.to_datetime(f'{year}-{month}-31')
    
    date_range = pd.date_range(start=start_date, end=end_date, freq="D")

    temp_df = pd.DataFrame(
        {
            "Date": date_range,
            "Debt": debt,
            "Percentage": percentage,
            "Debt_per_capita": debt_per_capita
        }
    )
    dfs.append(temp_df)
new_df = pd.concat(dfs, ignore_index=True)
new_df.drop_duplicates()
output_csv_path = "Debt_daily.csv"
new_df.to_csv(output_csv_path)
```
