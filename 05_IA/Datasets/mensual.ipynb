```python
import pandas as pd
import calendar
```

```python
file_path = 'gdp.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
df = df.drop(["TIME PERIOD"], axis=1)
df.head(100)
```

```python
dfs = []
```

```python
for index, row in df.iterrows():
    monthYear = row[0]
    euribor = row[1]
    year = int(monthYear[0:4])
    month = int(monthYear[5:7])

    is_leap_year = calendar.isleap(year)
    
    start_date = pd.to_datetime(f"{year}-{month}-01")

    if month == 4 or month == 6 or month == 9 or month == 11:
        end_date = pd.to_datetime(f'{year}-{month}-30')
    elif month == 2:
        if is_leap_year:
            end_date = pd.to_datetime(f'{year}-{month}-29')
        else:
            end_date = pd.to_datetime(f'{year}-{month}-28')
    else:
        end_date = pd.to_datetime(f'{year}-{month}-31')
    
    date_range = pd.date_range(start=start_date, end=end_date, freq="D")

    temp_df = pd.DataFrame(
        {
            "Date": date_range,
            "GDP_value": euribor
        }
    )
    dfs.append(temp_df)
new_df = pd.concat(dfs, ignore_index=True)
new_df.drop_duplicates()
output_csv_path = "GDP_daily.csv"
new_df.to_csv(output_csv_path, index=False)
```
