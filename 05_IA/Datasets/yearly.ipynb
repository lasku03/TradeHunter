```python
import pandas as pd
import calendar
```

```python
file_path = 'debt1.csv'
df = pd.read_csv(file_path)
dfs = []
df.head()
```

```python
df['Debt_GDP_percentage'] = df['Debt_GDP_percentage'] * 100
df.head()
```

```python
df['Debt_per_capita'] = df['Debt_per_capita'].str.replace('€', '').astype(float)
df['Debt_per_capita'] = df['Debt_per_capita'] * 1000
df.head()
```

```python
for index, row in df.iterrows():
    year = int(row[0])
    TotalDebt = row[1]
    Debt_percentage = row[2]
    per_capita = row[3]
    
    start_date = pd.to_datetime(f"{year}-01-01")
    end_date = pd.to_datetime(f"{year}-12-31")
    
    date_range = pd.date_range(start=start_date, end=end_date, freq="D")

    temp_df = pd.DataFrame(
        {
            "Date": date_range,
            "Total_debt": TotalDebt,
            "Debt_GDP_percentage": Debt_percentage,
            "Debt_per_capita": per_capita
        }
    )
    dfs.append(temp_df)
new_df = pd.concat(dfs, ignore_index=True)
new_df.drop_duplicates()
output_csv_path = "Debt.csv"
new_df.to_csv(output_csv_path,index=False)
```
