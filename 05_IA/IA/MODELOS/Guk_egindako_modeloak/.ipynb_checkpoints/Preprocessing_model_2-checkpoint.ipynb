#  Load Data

```python
import pandas as pd

df = pd.read_csv("archive_2/IBEX-1994-2020.csv")

# Ordena el DataFrame por la columna 'Date' para asegurarte de que las fechas estén en orden ascendente
df = df.sort_values(by='Date')


df.head()

df.columns
```

```python
import pandas as pd

# Supongamos que df es tu DataFrame original
# Asegurémonos de que la columna 'Date' esté en formato datetime
df['Date'] = pd.to_datetime(df['Date'])

# Agregar nuevas filas con datos específicos
nuevos_datos = df.append [
    {'Date': pd.to_datetime('2023-01-01'), 'Open': 100, 'High': 110, 'Low': 95, 'Close': 105, 'AdjClose': 105, 'Volume': 100000}
]

nuevos_datos.head()


# Imprimir el DataFrame actualizado
print(df)
```

```python
import pandas as pd

# Supongamos que df es tu DataFrame original
# Asegurémonos de que la columna 'Date' esté en formato datetime
df['Date'] = pd.to_datetime(df['Date'])

# Crea un rango de fechas que cubra desde la fecha mínima hasta la fecha máxima
rango_fechas = pd.date_range(start=df['Date'].min(), end=df['Date'].max(), freq='D')  # 'B' para días hábiles

# Reindexa el DataFrame usando el nuevo rango de fechas
df = df.set_index('Date').reindex(rango_fechas).reset_index()

# Si deseas llenar los valores faltantes con algún valor predeterminado, puedes hacerlo aquí
# Por ejemplo, llenar los valores de las columnas 'Open', 'High', 'Low', 'Close', 'AdjClose', y 'Volume' con ceros
columnas_numericas = ['Open', 'High', 'Low', 'Close', 'AdjClose', 'Volume']
df[columnas_numericas] = df[columnas_numericas].fillna(0)

# Si prefieres dejar los valores faltantes como NaN, puedes omitir la línea anterior

# Ordenar el DataFrame reindexado por la columna 'Date'
df = df.sort_values(by='Date')

# Imprimir el DataFrame ordenado
print("\nDataFrame ordenado por 'Date':")
print(df)
df.head()
```
