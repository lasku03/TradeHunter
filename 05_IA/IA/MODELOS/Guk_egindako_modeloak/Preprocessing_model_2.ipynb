#  Load Data

```python
import pandas as pd

datuak = pd.read_csv("archive_2/IBEX-1994-2020.csv")

# Ordena el DataFrame por la columna 'Date' para asegurarte de que las fechas estén en orden ascendente
datuak = datuak.sort_values(by='Date')

datuak.head()
```

```python
import pandas as pd

# Supongamos que datuak es tu DataFrame original
# Asegurémonos de que la columna 'Date' esté en formato datetime
datuak['Date'] = pd.to_datetime(datuak['Date'])

# Crea un rango de fechas que cubra desde la fecha mínima hasta la fecha máxima
rango_fechas = pd.date_range(start='1994-1-1', end=datuak['Date'].max(), freq='D')  # 'B' para días hábiles

# Reindexa el DataFrame usando el nuevo rango de fechas
datuak = datuak.set_index('Date').reindex(rango_fechas).reset_index()

datuak = datuak.fillna(0)
# Imprime el DataFrame actualizado
print(datuak)
```
