```python
import pandas as pd
```

```python
df = pd.read_csv("DJ_adapted.csv", parse_dates=['Date'])
df.head()
```

```python
df = df.drop(['Vol.', '% var.'], axis=1)
df.columns = ['']
df.head()
```

```python
df.sort_values('Date', inplace=True)
date_range = pd.date_range(start=df['Date'].min(), end=df['Date'].max())
df.head()
```

```python
df = df.set_index('Date').reindex(date_range).rename_axis('Date').reset_index()
df.head()
```

```python
df.fillna(method='ffill', inplace=True)
df.head()
```

```python
df.to_csv('DOW_JONES.csv', index=False)
```
