```python
import pandas as pd
```

```python
df = pd.read_csv("Datos históricos del IBEX 35 (1).csv")
```

```python
df = df.drop(['Vol.', '% var.'], axis=1)
df.head()
```

```python
df.columns = ['Date', 'Close', 'Open', 'High', 'Low']
```

```python
numeric_cols = ['Close', 'Open', 'High', 'Low']
for col in numeric_cols:
    # Remove periods (thousand separators)
    df[col] = df[col].str.replace('.', '')
    # Replace commas (decimal separators) with periods
    df[col] = df[col].str.replace(',', '.')
    # Convert to float
    df[col] = df[col].astype(float)
```

```python
df['Open'], df['High'], df['Low'], df['Close'] = df['High'], df['Low'], df['Close'], df['Open']
```

```python
df.columns = ['Date', 'Open', 'High', 'Low', 'Close']
```

```python
df['AdjClose'] = df['Close']
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d.%m.%Y')
```

```python
df.to_csv("IBEX-23.csv", index=False)
```
