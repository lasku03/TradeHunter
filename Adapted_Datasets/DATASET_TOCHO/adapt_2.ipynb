```python
import pandas as pd
```

```python
df = pd.read_csv("finished/Debt_finished.csv")
df.head()
```

```python
df = df.drop_duplicates(subset=['Date', 'Total_debt', 'Debt_GDP_percentage', 'Debt_per_capita'])
df.to_csv("Debt_finished.csv", index=False)
```

```python
df = df.drop(['Vol.', '% var.'], axis=1)
df.head()
```

```python
df.columns = ['Date', 'EUR_Price', "EUR_Open", "EUR_High", "EUR_Low"]
df.head()
```

```python
numeric_cols = ['EUR_Price', 'EUR_Open', 'EUR_High', 'EUR_Low']
for col in numeric_cols:
    # Remove periods (thousand separators)
    df[col] = df[col].str.replace('.', '')
    # Replace commas (decimal separators) with periods
    df[col] = df[col].str.replace(',', '.')
    # Convert to float
    df[col] = df[col].astype(float)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df = df.sort_values(by='Date')
df.head()
```

```python
df['Open'], df['High'], df['Low'], df['Close'] = df['High'], df['Low'], df['Close'], df['Open']
```

```python
df.columns = ['Date', 'Open', 'High', 'Low', 'Close']
```

```python
df['AdjClose'] = df['Close']
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d.%m.%Y')
```

```python
df.to_csv("EUR-2023.csv", index=False)
```
