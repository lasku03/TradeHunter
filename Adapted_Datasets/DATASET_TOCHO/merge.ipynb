```python
import pandas as pd

ibex_data = pd.read_csv('mergedataset.csv')
dow_jones_data = pd.read_csv('finished/IPC_finished.csv')

ibex_data['Date'] = pd.to_datetime(ibex_data['Date'])
dow_jones_data['Date'] = pd.to_datetime(dow_jones_data['Date'])

merged_data = pd.merge(dow_jones_data, ibex_data, on='Date', how='inner')

merged_data.to_csv('mergedataset.csv', index=False)

merged_data.head()
```
